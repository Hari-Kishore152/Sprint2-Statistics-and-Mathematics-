### 1. Introduction to Probability

Probability is a mathematical way of measuring how likely or unlikely an  Event is to happen.

The value of probability is always between: 0 and 1

- where 0 means impossible.
- 0.5	Equal chance of happening or not happening.
- 1 means certain.

Probability = Likelihood of an event

**Formula**
  
**When all outcomes are equally likely:**

**𝑃(A)=  Number of favourable outcomes / Total number of possible outcomes**

Consider rolling a fair six-sided dice.

Possible outcomes: {1, 2, 3, 4, 5, 6}

Probability of getting 4 : P(4)=1/6


P(A) = probability of event A
favorable outcomes = outcomes that satisfy what we're looking for
total possible outcomes = all possible outcomes


```python
favorable_outcomes = 3
total_outcomes = 6

probability = favorable_outcomes / total_outcomes

print("Probability:", probability)
```

### 2. Events

An event in probability is a specific outcome or a group of outcomes from a random experiment.

Mathematically, an event is a subset of the sample space

A **random experiment** is an activity where we know the possible outcomes, but we don't know exactly which outcome will occur before performing it.

Event = A specific outcome or set of outcomes of a random experiment.

**Example: rolling a die**

When you roll a six-sided die, the sample space is:
    
       s={1,2,3,4,5,6}

An event can be: 
Event A = Rolling an even number   A={2,4,6}


Another event can be:
Event B = Rolling a number greater than 4    B={5,6}

Each event contains the outcomes that satisfy a particular condition.

```python
sample_space = {1, 2, 3, 4, 5, 6}

# Event A: rolling an even number
event_A = {2, 4, 6}

probability_A = len(event_A) / len(sample_space)

print("Event A:", event_A)
print("Probability of A:", probability_A)
```

### 3. Sample Space

A sample space is the set of all possible outcomes of a random experiment. It is usually represented by 
𝑆  or  Ω

**Examples**
  
Tossing one coin
Possible outcomes are:    Head or Tail

So the sample space is:   S= {H,T}

Tossing two coins
Each toss can be Head or Tail :   S={HH,HT,TH,TT}

Rolling a die: {1, 2, 3, 4, 5, 6}



```python
import itertools

coin = ["H", "T"]

# Generate all possible outcomes for two coin tosses
sample_space = list(itertools.product(coin, repeat=2))

print("Sample space:", sample_space)
print("Number of outcomes:", len(sample_space))

# Exactly one head
event_A = [outcome for outcome in sample_space if outcome.count("H") == 1]

print("Exactly one head:", event_A)
print("Probability:", len(event_A) / len(sample_space))
```

### 4.Independent Events


Independent events are events where the occurrence of one event does not affect the probability of the other event occurring.


This means that even after knowing that A happened, the probability of B remains unchanged.

**Formula**

For independent events A and B:
                P(A∩B) = P(A) P(B)

P(A) = probability of event 

P(B) = probability of event 

P(A∩B) = probability that both 𝐴 and B happen

    
**example**
    
**Suppose we toss a coin twice.**

*First toss:*     Heads

    Does getting Heads on the first toss change the probability of getting Heads on the second toss?

     No.

The second toss is a new toss.

Even if the first toss was Heads:

P(Heads on second toss) = 0.5
    
Therefore, the two events are independent.

Two events are Independent if the occurrence of one does not affect the occurrence of the other.

**another Example:**

Roll a dice and toss a coin.

The dice result does not affect the coin result.

```python
p_A = 0.5
p_B = 0.5

# Probability that both independent events occur
p_both = p_A * p_B

print("Probability of both heads:", p_both)
```

### 5. Dependent Events

Dependent events are events in which the occurrence of one event changes the probability of the other event occurring

What happens in the first event affects what can happen in the second event.

**Formula**

For dependent events A and 𝐵 :
         $P(A∩B)=P(A)×P(B∣A)$


P(A) = probability that event A occurs

P(B∣A) = probability that event B occurs given that A has already occurred

P(A∩B) = probability that both events occur


Dependent events satisfy:
$P(B∣A)!=P(B)$






**Imagine a bag containing:**

3 red balls
2 blue balls

**Total:**  5 balls

- You pick one ball without putting it back.

   Suppose you pick a red ball.

- Now the bag contains:

2 red ,2 blue

**There are now only 4 balls.**

**The probability of picking red on the second draw has changed.**

*Before first draw:*

$$ P(Red)=\frac{3}{5} $$

*After drawing a red:*

$$ P(Red)=\frac{2}{4} $$

The probability changed.

Therefore, the events are dependent.



First red:

$$ P(A)=\frac{3}{5} $$

Second red given first red:

$$ P(B|A)=\frac{2}{4} $$

Therefore:

$$ P(A\cap B) = \frac{3}{5}\times\frac{2}{4} $$ $$ =\frac{6}{20} $$ $$ =0.30 $$

So there is a 30% probability of drawing two red balls in a row without replacement.

```python
p_first_red = 3 / 5
p_second_red = 2 / 4

probability = p_first_red * p_second_red

print("Probability of two red balls:", probability)
```

### 6. Conditional Probability

Conditional probability is the probability that an event occurs given that we already know another event has occurred

or
    
Conditional probability is the probability of an event occurring when we know that another event has occurred.
    
It tells us how the likelihood of one event changes when we know something about another event.

**Formula**
    
$$ P(A|B)=\frac{P(A\cap B)}{P(B)} $$

where:

P(A|B) = probability of A given B

P(A ∩ B) = probability that A and B both occur

P(B) = probability of B

**Detailed Explanation**

Suppose we ask:

What is the probability that a customer purchases a product?

Maybe we know that overall:

P(Purchase) = 20%

But now we receive additional information:

The customer clicked the advertisement.

Maybe customers who clicked the advertisement are much more likely to purchase.

So we ask:

What is the probability of purchase given that the customer clicked the advertisement?

That's conditional probability.

```python
p_a_and_b = 0.30
p_b = 0.60

conditional_probability = p_a_and_b / p_b

print("P(A | B):", conditional_probability)
```

### 7. Bayes' Theorem

Bayes' Theorem is a formula that relates conditional probabilities, allowing you to update the probability of a hypothesis when new evidence is observed. 

It is written as:
P(A∣B)= P(B∣A)P(A)/P(B)

Bayes' Theorem helps us update the probability of an event when new evidence becomes available.

Bayes' theorem helps us update what we believe after seeing new information.

example:

P(Disease)=0.01

P(Positive|Disease)=0.99

P(Positive)=0.05

Then

P(Disease|Positive)

=(0.99×0.01)/0.05

=0.198



```python
P_A = 0.01
P_B_given_A = 0.99
P_B = 0.05

bayes = (P_B_given_A * P_A) / P_B

print(bayes)
```

### 8. Probability Rules

There are several basic probability rules that you should know.


**Rule 1 – Probability is Between 0 and 1**

Every probability must satisfy:

$$ 0\leq P(A)\leq1 $$
Explanation

Probability cannot be negative.

It also cannot be greater than 1.

0     → 0%

0.25  → 25%

0.50  → 50%

0.90  → 90%

1     → 100%

**Rule 2 – Complement Rule**

The complement of an event means the event does not happen.

Complement Rule (The "Not" Rule)Used to find the probability that an event will not happen.

**Rule 3 – Addition Rule**

The addition rule is used when we want the probability of:  A or B

Addition Rule (The "Or" Rule) Used to find the probability that either event [A] or event [B] occurs

P(A∪B)=P(A)+P(B)−P(A∩B)

**Rule 4 – Multiplication Rule**

Multiplication Rule (The "And" Rule) Used to find the probability that both event [A] and event [B] occur together

**Rule 5 – Total Probability**

If several mutually exclusive events cover all possible cases:

$$ P(B)= P(B|A_1)P(A_1) + P(B|A_2)P(A_2) +\cdots $$

This rule is especially useful when applying Bayes' theorem.

### 9. Random Variables

A random variable is a variable whose value is determined by the outcome of a random experiment.

A numerical way of representing the outcome of a random experiment.

A random variable assigns a numerical value to each possible outcome of a random experiment.

Two common types are:

discrete random variable: takes countable values, such as the number of heads in three coin tosses.

Continuous random variable: can take values across a continuous range, such as waiting time or a person's heigh

Suppose we toss a coin.

Outcomes:

Heads
Tails

We can define a random variable X:

X = 1 if Heads
X = 0 if Tails

Now the random variable converts the outcome into a numerical value.

So:

Heads → X = 1
Tails → X = 0

```python
import itertools
from collections import Counter

coin = ["H", "T"]

outcomes = list(itertools.product(coin, repeat=2))

# X = number of heads
head_counts = [outcome.count("H") for outcome in outcomes]

frequencies = Counter(head_counts)
total = len(outcomes)

for value in sorted(frequencies):
    probability = frequencies[value] / total
    print(f"P(X = {value}) = {probability}")
```

### 10. Expected Value

The expected value is the long-run average outcome we would expect if a random experiment were repeated many times under the same conditions

```python
outcomes = [100, 0]
probabilities = [0.2, 0.8]

expected_value = sum(
    value * probability
    for value, probability in zip(outcomes, probabilities)
)

print("Expected payout: ₹", expected_value)
```
