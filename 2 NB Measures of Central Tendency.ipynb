## Notebook 2 – Measures of Central Tendency


**Definition**

**Measures of central tendency** are statistical methods used to find a central, typical, or representative value in a dataset.

They help us understand the overall pattern of the data.

Central tendency helps us represent many values using one value that gives us an idea about where the data is generally centered.

- The three most commonly used measures are:

    Mean

    Median

    Mode

### 1. Mean

**definition**

In statistics, the mean is one of the most common measures of central tendency. 

It represents the average value of a dataset and gives a sense of the “typical” number around which the data points cluster.

The mean tells us roughly where most of the values in the dataset are concentrated.

**Mean = sum of all values ÷ number of values.**

The formula for the arithmetic mean is:

- x represents an individual value in our dataset.
- ∑x  sigma adds all the values together
- n represents the number of observations/values.

$$
\text{Mean} = \frac{\sum_{i=1}^{n} x_i}{n}
$$


**Example**

- Suppose an ML dataset contains the ages of customers:

22, 25, 28, 31, 34

- We can calculate the average age.

```python

import numpy as np

ages = np.array([22, 25, 28, 31, 34])

average_age = np.mean(ages)

print("Average age:", average_age)


# This can be useful during Exploratory Data Analysis (EDA).

# We may want to understand:

# "What is the typical age of customers in this dataset?"

# The mean gives us one quick summary.
```

### 2. Median

The median is the middle value of a dataset after the values have been arranged in ascending or descending order.In simple words:

Median tells us the value that lies in the middle of the ordered data.

For example:

10, 20, 30, 40, 50

The middle value is:

30

- Case 1: Odd number of observations:

     If there are an odd number of values:
        Median = middle value

  
- Case 2: Even number of observations

   If there are an even number of values, there are two middle values.

   We calculate their average.

Example:

10, 20, 30, 40

The two middle values are:

20 and 30

Median=  20+30/2

Median = 25


\[
\text{Median} =
\begin{cases}
x_{\left(\frac{n+1}{2}\right)}, & \text{if } n \text{ is odd} \\
\frac{x_{\left(\frac{n}{2}\right)} + x_{\left(\frac{n}{2}+1\right)}}{2}, & \text{if } n \text{ is even}
\end{cases}
\]

```python
import numpy as np

marks = np.array([10, 20, 30, 40, 50])

median = np.median(marks)

print("Median:", median)
```

```python
numbers = np.array([10, 20, 30, 40])

median = np.median(numbers)

print("Median:", median)
```

### 3. Mode

The mode is the value that occurs most frequently in a dataset.

In simple words:
Mode tells us which value appears the most number of times.

A dataset can have:

- One mode → unimodal

- Two modes → bimodal

- More than two modes → multimodal

Example:

10, 20, 20, 30, 40

20 appears twice.

The other values appear once.

Therefore:

Mode = 20

Important Point:

Mode is different from mean and median.

- Mean → average
- Median → middle value
- Mode → most frequent value

$$
\text{Mode} = \text{Value that occurs most frequently in the dataset}
$$


```python
from scipy import stats

numbers = [2, 3, 3, 4, 5, 3, 6]

mode = stats.mode(numbers)

print("Mode:", mode.mode)
```

### 4. Weighted Mean

Sometimes, all values do not have equal importance.

In such situations, a normal mean may not be appropriate.

A weighted mean allows some values to contribute more than others.

**definition**

A weighted mean (or weighted average) is an average in which some values are given more importance than others by assigning them weights. Instead of treating every value equally, values with higher weights affect the final average more

Example: exam marks

Suppose your final grade is calculated like this:

$$
\text{Weighted Mean} = \frac{\sum_{i=1}^{n} w_i \cdot x_i}{\sum_{i=1}^{n} w_i}
$$


```python
import numpy as np

marks = np.array([80, 70, 90])
weights = np.array([20, 30, 50])

weighted_mean = np.average(marks, weights=weights)

print(weighted_mean)
```

### 5. Geometric Mean

The geometric mean is a type of average calculated by multiplying all positive values together and then taking the n-th root of the product,
where n is the number of values.

Why is Geometric Mean useful?

Geometric mean is particularly useful when dealing with:

- growth rates
- investment returns
- ratios
- percentages
- multiplicative changes

It is generally more appropriate than arithmetic mean when values represent growth over multiple periods.

$$
\text{Geometric Mean} = \sqrt[n]{x_1 \cdot x_2 \cdot \dots \cdot x_n}
$$


```python
#Python Implementation

import numpy as np

growth_factors = np.array([1.10, 1.20])

geometric_mean = np.prod(growth_factors) ** (1 / len(growth_factors))

average_growth = (geometric_mean - 1) * 100

print("Average growth rate:", average_growth, "%")
```

```python
#Using SciPy

from scipy.stats import gmean

growth_factors = [1.10, 1.20]

geometric_mean = gmean(growth_factors)

average_growth = (geometric_mean - 1) * 100

print("Average growth rate:", average_growth, "%")
```

### 6.Harmonic mean

The harmonic mean is a type of average calculated by dividing the total number of values by the sum of their reciprocals. It is mainly used for averaging rates, ratios, speeds, and time-per-unit values.


$$
\text{Harmonic Mean} = \frac{n}{\sum_{i=1}^{n} \frac{1}{x_i}}
$$


It is especially useful when dealing with:

- rates
- ratios
- speeds
- rates where the denominator matters

**Simple example**
  
Find the harmonic mean of 20 and 60:

$$
\text{HM} = \frac{2}{\frac{1}{20} + \frac{1}{60}}
$$

$$
\text{HM} = \frac{2}{\frac{3}{60} + \frac{1}{60}}
$$

$$
\text{HM} = \frac{2}{\frac{4}{60}}
$$

$$
\text{HM} = \frac{2 \cdot 60}{4} = 30
$$


```python
#Python Implementation
speeds = [60, 40]

harmonic_mean = len(speeds) / sum(1 / speed for speed in speeds)

print("Harmonic Mean:", harmonic_mean)
```

```python
# Using SciPy
from scipy.stats import hmean

speeds = [60, 40]

result = hmean(speeds)

print("Harmonic Mean:", result)
```
