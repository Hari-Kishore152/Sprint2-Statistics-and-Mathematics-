### 1.Introduction to Statistics

**DEFINITION**

Statistics is the branch of applied mathematics that deals with collecting, organizing, analyzing, interpreting, and presenting data to identify patterns and support decision‑making.

- Data by itself may not tell us much. Statistics gives us methods to convert that raw data and figures into meaningful information.
- It helps us understand large populations by studying smaller samples,
- Used to make decisions when studying an entire population is impractical.
- It involves data collection, description, analysis, and inference.

**Why it matters**

Statistics helps people test ideas, estimate risks, identify trends, and make evidence-based decisions rather than relying on guesswork. It is widely used in medicine, business, government, science, finance, and data science

**Types of Statistics:**

- Descriptive Statistics ->  Summarizes and describes data using measures like mean, median, mode, variance, and graphs.

     EXAMPLE: Calculating the average score of a class or making a bar chart

- Inferential Statistics ->  Uses sample data to make predictions or generalizations about a population.

     EXAMPLE: Surveying 1,000 voters to estimate how all voters will vote

**EXAMPLE:**

For example, imagine that we have the marks of 1,000 students. Looking at 1,000 individual marks one by one would be difficult. Statistics allows us to summarize those marks using values such as:

- Mean
- Median
- Minimum
- Maximum
- Range
- Variance
- Standard deviation

**These basic statistical calculations are commonly used during Exploratory Data Analysis (EDA) to understand a dataset before training a machine learning model.**

Statistics also helps us identify patterns and relationships in data.

It can help answer questions such as:

- What is the average value?
- How much does the data vary?
- What is the most common value?
- Are there unusual values?
- Are two variables related?
- Can we use a smaller group to understand a larger group?

So, we can think of statistics as a set of methods that helps us make sense of data.

```python
import numpy as np

marks = np.array([60, 70, 80, 90, 100])   #here we create an numpy array containing the marks of five students.

mean = np.mean(marks)                                 #np.mean() calculates the average.
minimum = np.min(marks)                               #np.min() finds the smallest value.
maximum = np.max(marks)                               #np.max() finds the largest value.
data_range = np.max(marks) - np.min(marks)            #this calculates the range

print("Marks:", marks)
print("Mean:", mean)
print("Minimum:", minimum)
print("Maximum:", maximum)
print("Range:", data_range)
```

### 2. Why Statistics is Important in AI/ML

Statistics is important in AI and Machine Learning because machine learning models depend on data, and statistics helps us understand that data.

A machine learning model cannot be treated as a black box where we simply provide data and expect meaningful results. Before training a model, we need to understand the quality and behavior of the data.

Statistics is important in AI and Machine Learning because AI/ML systems learn patterns from data, and statistics gives us the tools to understand that data, handle uncertainty, build models, and judge whether their predictions are reliable. 

In simple words: statistics helps machines learn from data instead of guessing.

Without statistics, ML algorithms cannot generalize, detect patterns, or avoid errors like overfitting.


**Statistics is important in AI/ML because it helps us:**

- 1.Understand and clean data  : Statistics helps identify missing values, incorrect entries, duplicates, and unusual outliers before training a model.

- 2.Discover patterns and relationships  : It helps find how variables are connected, such as whether more study hours are related to higher exam marks.

- 3.Measure probability and uncertainty  : It enables models to estimate likelihoods and confidence, such as “there is an 85% chance this transaction is fraudulent.”

- 4.Select useful features  : Statistical analysis helps choose the input columns that meaningfully affect the prediction and remove irrelevant ones.

- 5.Build statistical and ML models  : Many ML algorithms use statistical concepts to learn patterns from historical data and make predictions on new data.

- 6.Evaluate model accuracy and errors  : Metrics such as accuracy, precision, recall, MAE, and RMSE help measure how well a model performs.

- 7.Avoid overfitting  : Statistical validation methods help ensure a model learns general patterns rather than memorizing the training data.

- 8.Make more trustworthy decisions  : Statistics helps explain the reliability, risk, confidence, and possible error behind an AI model’s prediction.



### 3. Statistics vs Machine Learning

Statistics and Machine Learning are closely related because both work heavily with data, but their primary purposes can be different.

Statistics is commonly concerned with understanding data, relationships, estimating unknown quantities, and drawing conclusions from data  while accounting for uncertainty.

Machine Learning is commonly concerned with learning patterns from data and using those patterns to make predictions or decisions on new data.

**Statistics explains data; Machine Learning predicts from data.**

**They overlap a lot, and statistics is one of the main foundations of machine learning.**




**Statistics asks:**  Does studying more hours actually relate to higher marks?

A statistical analysis might calculate correlation, fit a regression line, test a hypothesis, and estimate how strongly study hours influence marks.

Example conclusion:

There is a strong positive relationship between study hours and exam marks

**Machine Learning asks:**  If a new student studies 5 hours, what mark might they get?

A trained ML model might predict:
Predicted mark: 68

So, statistics explains the relationship; machine learning uses learned patterns to make predictions.

**example:**

A statistical analysis might ask:

     "Is there a relationship between experience and salary?"

A machine learning problem might ask:

     "Can we predict the salary of an employee with 6 years of experience?"

The same data can therefore be used for both statistical analysis and machine learning.

Simple Difference:

Statistics
    ↓
Understand / analyze / estimate


Machine Learning
    ↓
Learn patterns
    ↓
Predict / classify / make decisions

There is significant overlap between the two fields.**Machine learning uses many statistical concepts.**


**Suppose we have:**

Experience = 1, 2, 3, 5, 8
Salary     = 30, 35, 40, 55, 80

A statistical approach can calculate the correlation between experience and salary.

A machine learning approach can use the same data to train a model that predicts salary.

So the data can be the same, but the question we ask can be different.

```python
#This code does two things with the same data: first it finds the statistical relationship between experience and salary, 
#then it trains a machine-learning linear regression model to predict salary

import numpy as np
from sklearn.linear_model import LinearRegression

experience = np.array([1, 2, 3, 5, 8])
salary = np.array([30, 35, 40, 55, 80])

# Statistical analysis
correlation = np.corrcoef(experience, salary)[0, 1]    #This line calculates the correlation coefficient between experience and salary.

print("Correlation:", correlation)

# Machine Learning
X = experience.reshape(-1, 1)

model = LinearRegression()
model.fit(X, salary)

predicted_salary = model.predict([[6]])

print("Predicted salary for 6 years of experience:", predicted_salary[0])
```

### 5.Inferential Statistics

**definitioin**

Inferential statistics is the branch of statistics that uses data from a sample to make conclusions, estimates, or predictions about a larger population. 

This is important because studying an entire population is often difficult or expensive.

**Inferential Statistics** = Using sample data to draw conclusions or learn about an entire population.

For example, suppose a company has 100,000 customers.

Instead of asking every customer about their satisfaction, the company may select 1,000 customers.

The 1,000 customers are the sample.

The 100,000 customers are the population.

**Important Point:** Inferential statistics involves uncertainty.

A sample may not perfectly represent the entire population.

Therefore, inferential statistics uses probability and other statistical methods to quantify or account for this uncertainty.

```python
import numpy as np

# Population
population = np.random.normal(
    loc=170,
    scale=8,
    size=10000
)

# Select a sample
sample = np.random.choice(
    population,
    size=500,
    replace=False
)

population_mean = np.mean(population)
sample_mean = np.mean(sample)

print("Actual Population Mean:", population_mean)
print("Sample Mean:", sample_mean)


# Code Explanation:

# We first create a population containing 10,000 values.

# Then we randomly select 500 values from that population.

# Finally, we calculate:

# np.mean(population)

# and:

# np.mean(sample)
```

Interpretation:

The sample mean should generally be reasonably close to the population mean when the sample is selected appropriately, 
although it will not necessarily be exactly the same.

**Where This Is Used in AI/ML:**
Sampling is common when datasets are extremely large. 
Statistical reasoning helps us understand whether a sample is representative and how much uncertainty may exist in conclusions based on it.

### 7. Sample

**definition**

In statistics, a sample is a subset of the population that you actually collect and analyze data from. Since the full population is often too large or impossible to measure, we use samples to make inferences about the population.

The population represents the complete group we are interested in.

The sample represents the smaller group we actually study.

Researchers use the sample to estimate or draw conclusions about the whole population.

In AI/ML, the sample is your dataset — the portion of real-world data you use to train, validate, and test models.

A good sample **should be representative**, meaning it should reflect the important characteristics of the larger population as fairly as possible.

```python
import numpy as np

population = np.arange(1, 10001)

sample = np.random.choice(
    population,
    size=100,
    replace=False
)

print("Population Size:", len(population))
print("Sample Size:", len(sample))
print("Sample:", sample)
```

**Where This Is Used in AI/ML**

Sampling can be used when working with very large datasets where analyzing or processing every observation at every stage may be expensive.

### 8. Census

In a census, we do not select only a sample.

We attempt to collect information from the entire population

It provides the most accurate picture of the population but is often costly, time‑consuming, and sometimes impractical.

**For example:**

suppose a company has 5,000 employees.

### 9. Sampling

**definition**

In statistics, sampling is the process of selecting a subset of individuals, items, or data points (called a sample) from a larger population to study and analyze. Since it’s often impractical to collect data from the entire population (a census), sampling allows us to make inferences about the population using manageable data.

Sampling : the method of choosing a representative portion of the population.
Goal: Ensure the sample reflects the characteristics of the population to avoid bias.

- Population → the complete group

- Sample → the selected smaller group

- Sampling → the process of selecting that smaller group

Sampling is the bridge between population and dataset. In AI/ML, your dataset is a sample, and the quality of your model depends on how well that sample represents the population.

For example:

Population = 10,000 customers

Sample = 500 customers

Sampling = The process used to select those 500 customers

**Why Is Sampling Important?**

The quality of our statistical conclusion depends heavily on how the sample was selected.

A sample should ideally represent the population.

For example, suppose a college has:

50% Male

50% Female

If we randomly select a sample of 500 students, we would generally want the sample to reasonably reflect the population rather than selecting students from only one group.

**Types of Sampling:**
    
- Random Sampling → Each member has equal chance of selection.

- Stratified Sampling → Population divided into groups (strata), then sampled proportionally.

- Cluster Sampling → Population divided into clusters, some clusters chosen randomly.

- Systematic Sampling → Selecting every kth item from a list.

- Convenience Sampling → Choosing data that’s easiest to access (often biased).

```python
import numpy as np

population = np.arange(1, 10001)

sample_size = 500

sample = np.random.choice(
    population,
    size=sample_size,
    replace=False
)

print("Population Size:", len(population))
print("Sample Size:", len(sample))
print("First 20 Sample Values:", sample[:20])


# Code Explanation

# We first create a population containing 10,000 observations.

# Then:

# np.random.choice()

# randomly selects 500 observations.

# replace=False

# ensures that an observation cannot be selected twice.
```

**Where This Is Used in AI/ML**

Sampling is useful when datasets are very large.

### 11. Features

In Machine Learning, a feature is an input variable or measurable piece of information that a model uses to make a prediction. Features are also called predictors, independent variables, or input columns

These are the features because the ML model uses them as inputs to predict the result.

The model uses the features to learn patterns that help predict the target.

If our goal is to predict Price, then:

- Area
- Bedrooms
- Age
- Location
These are the features because the ML model uses them as inputs to predict the result.

Price is the value we are trying to predict, so it is commonly called the target or target variable.

```python
import pandas as pd

data = {
    "Study_Hours": [2, 4, 6, 8, 10],
    "Attendance": [60, 70, 80, 90, 95],
    "Previous_Marks": [45, 55, 65, 75, 85],
    "Result": ["Fail", "Fail", "Pass", "Pass", "Pass"]
}

df = pd.DataFrame(data)

print(df)
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Feature / input variable (X)
# Each inner list represents one student
study_hours = np.array([                       #study_hours is the feature because it is the input the model uses to predict marks.
    [1],
    [2],
    [3],
    [4],
    [5]
])

# Target / output variable (y)
marks = np.array([40, 50, 60, 70, 80])

# Create the machine learning model
model = LinearRegression()

# Train the model
model.fit(study_hours, marks)

# Predict marks for a student who studies 6 hours
predicted_marks = model.predict([[6]])

print("Predicted marks:", predicted_marks[0])
```
