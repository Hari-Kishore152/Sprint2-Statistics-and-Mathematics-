## Notebook 4 – Data Distribution

**What is a Data Distribution?**

Before learning the individual distributions, let's understand what distribution means.

Definition

A data distribution describes how the values in a dataset are arranged or spread across different possible values.

A distribution tells us how frequently different values occur and how the data is shaped or spread.

### 1. Normal Distribution

The normal distribution (also called the Gaussian distribution) is a continuous probability distribution that describes data symmetrically around the mean, forming the familiar “bell curve.” It is defined by two parameters: the mean (μ), which sets the center, and the standard deviation (σ), which controls the spread.

It is a continuous probability distribution that is:

symmetric around its mean,
centered around the mean,
highest near the mean,
and gradually decreases as we move away from the mean.

Its shape looks approximately like a bell.

- Normal Distribution = A probability distribution where most values cluster around the mean, and fewer values appear as you move further away.

- Shape = Bell‑shaped, symmetric about the mean.

Parameters:

Mean (μ): Center of the distribution.

Standard Deviation (σ): Controls how spread out the data is.

**Important properties**

For a perfectly normal distribution:

$$ Mean = Median = Mode $$

For example:

Mean   = 50
Median = 50
Mode   = 50

The distribution is also symmetric.

That means the left side is approximately a mirror image of the right side.

```python
import numpy as np
import matplotlib.pyplot as plt

# Sample marks of 100 students
marks = np.random.normal(loc=70, scale=10, size=100)

plt.hist(marks, bins=10, edgecolor="black")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")
plt.show()
```

**The probability density function (PDF) of the normal distribution is:**

$$
f(x) = \frac{1}{\sigma \sqrt{2\pi}} \, e^{-\frac{1}{2}\left(\frac{x-\mu}{\sigma}\right)^2}
$$


$$
f(x) = \text{probability density at value } x
$$

$$
\mu = \text{mean}
$$

$$
\sigma = \text{standard deviation}
$$

$$
e = \text{Euler's number } (\approx 2.718)
$$

$$
\pi = \text{pi } (\approx 3.14159)
$$


```python
import numpy as np
import matplotlib.pyplot as plt

mean = 50
std = 10

x = np.linspace(10, 90, 500)

pdf = (
    1 / (std * np.sqrt(2 * np.pi))
    * np.exp(-((x - mean) ** 2) / (2 * std ** 2))
)

plt.plot(x, pdf)
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.title("Normal Distribution")
plt.show()
```

### 2. Uniform Distribution

A uniform distribution is a probability distribution in which all possible outcomes are equally likely within a specified range. In other words, no value is favored over another.

Uniform distribution = A probability distribution where all outcomes, or equal-sized intervals, have equal probability.

or

A Uniform Distribution is a distribution where every value within a given range has an equal chance of occurring.

Shape: Flat, constant height between 𝑎 and 𝑏


**For a continuous uniform distribution between \(a\) and \(b\):**

\[
f(x) =
\begin{cases}
\frac{1}{b-a}, & a \leq x \leq b \\
0, & \text{otherwise}
\end{cases}
\]

Where:

a = lower limit
b = upper limit

a= minimum value

𝑏= maximum value

f(x) = constant probability density between 𝑎and 𝑏


**Example Using the Formula:**

Suppose:

a = 10
b = 20

$$
f(x) = \frac{1}{20 - 10}
$$

$$
= \frac{1}{10}
$$

$$
= 0.1
$$


So the probability density is:

0.1

```python
import numpy as np
import matplotlib.pyplot as plt

a = 10
b = 20

x = np.linspace(5, 25, 500)

pdf = np.where(
    (x >= a) & (x <= b),
    1 / (b - a),
    0
)

plt.plot(x, pdf)
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.title("Uniform Distribution")
plt.show()
```

### 3.Gaussian Distribution

Gaussian distribution and normal distribution refer to the same probability distribution.

The name Gaussian comes from the mathematician Carl Friedrich Gauss.

It has:

- a bell-shaped curve,
- symmetry,
- mean = median = mode,
- a center determined by μ,
- spread determined by σ.

```python
import numpy as np
import matplotlib.pyplot as plt

mu = 100
sigma = 15

x = np.linspace(40, 160, 500)

pdf = (
    1 / (sigma * np.sqrt(2 * np.pi))
    * np.exp(-((x - mu) ** 2) / (2 * sigma ** 2))
)

plt.plot(x, pdf)
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.title("Gaussian Distribution")
plt.show()
```

### 4. Symmetric Distribution

A distribution is symmetric when the left and right sides have approximately the same shape around the center.

- For a perfectly symmetric distribution:

    Skewness=0

- For a symmetric distribution with one peak, usually:

    Mean=Median=Mode


or

A symmetric distribution is one where the left and right sides of the distribution are mirror images around a central point. In such distributions, the mean, median, and mode all coincide at the center, and skewness is zero

- Values: 60, 70, 80, 90, 100

The center is 80.

The values are equally spaced around 80:

- 60 and 100 are equally far from 80.

- 70 and 90 are equally far from 80.

So this is a symmetric dataset.

```python
import numpy as np
import matplotlib.pyplot as plt

data = [2, 4, 6, 8, 10]

print("Mean:", np.mean(data))
print("Median:", np.median(data))

plt.hist(data, bins=5, edgecolor="black")
plt.title("Example of Symmetric Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

### 5.Left Skew / Negative Skew

A distribution is left-skewed when its longer tail extends toward the left side.

A left-skewed distribution is a distribution where the long tail extends toward the left, or toward lower values.

Most data values are concentrated on the right side (higher values), while a few unusually low values pull the tail and the mean to the left.

A Left-Skewed Distribution has a long tail on the left side.

Most observations are concentrated on the right side.

Tail direction: Extends to the left (negative side).

Skewness: Negative.

Mean, Median and Mode

**In a typical left-skewed distribution:**

      Mean<Median<Mode


**Examples:**

Age at death in a wealthy country (most people live long, but a few die young).

Exam scores where most students score high, but a few score very low.

```python
import numpy as np
from scipy.stats import skew

data = np.array([
    20, 45, 50, 55, 60,
    62, 65, 67, 68, 70
])

skewness = skew(data)

print("Skewness:", skewness)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Most values are high, with a smaller number of low values
data = [1, 7, 8, 8, 9, 9, 9, 10, 10, 10, 10]

print("Mean:", np.mean(data))
print("Median:", np.median(data))

plt.hist(data, bins=10, edgecolor="black")
plt.title("Left-Skewed Data Example")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

```python
left_skew = np.random.beta(5, 2, 100) * 100

plt.hist(left_skew, bins=10, edgecolor="black")
plt.title("Left Skewed Distribution")
plt.show()
```

### 6. Right Skew / Positive Skew

A right-skewed distribution is a distribution in which the long tail extends toward the right, or toward higher values. It is also called a positively skewed distribution

Most values are concentrated on the left side (lower values), while a few unusually high values pull the distribution and mean to the right.

**Mean, Median and Mode**

Typically:

$$ Mode < Median < Mean $$

The high extreme values pull the mean toward the right.

A Right-Skewed Distribution has a long tail on the right side.

Most observations are concentrated on the left side.

A few unusually large values can pull the mean toward the right tail.

```python
import numpy as np
from scipy.stats import skew

data = np.array([
    10, 12, 15, 18, 20,
    22, 25, 30, 100
])

skewness = skew(data)

print("Skewness:", skewness)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Many small values and a few large values
spending = [10, 12, 15, 18, 20, 22, 25, 30, 35, 200]

print("Mean:", np.mean(spending))
print("Median:", np.median(spending))

plt.hist(spending, bins=10, edgecolor="black")
plt.title("Right-Skewed Customer Spending")
plt.xlabel("Spending")
plt.ylabel("Frequency")
plt.show()
```

```python
right_skew = np.random.beta(2, 5, 100) * 100

plt.hist(right_skew, bins=10, edgecolor="black")
plt.title("Right Skewed Distribution")
plt.show()
```

### 7. Kurtosis

Kurtosis is a statistical measure that describes how heavy or light the tails of a distribution are compared with a normal distribution. It helps indicate how likely a dataset is to contain extreme values or outliers.

High kurtosis → More extreme values / heavier tails
Low kurtosis → Fewer extreme values / lighter tails

- Mesokurtic =	0	Similar tail heaviness to a normal distribution
- Leptokurtic =	Greater than 0	Heavy tails; more chance of extreme values/outliers
- Platykurtic =	Less than 0	Light tails; fewer extreme values/outliers

```python
import numpy as np
from scipy.stats import kurtosis

data = [10, 12, 13, 14, 15, 16, 17, 18, 20, 100]

print("Excess kurtosis:", kurtosis(data))
```

```python
import pandas as pd

data = pd.Series([10,12,14,15,16,18,20])

print(data.kurt())
```

```python
import numpy as np
from scipy.stats import kurtosis

data = np.array([
    10, 12, 15, 18, 20,
    22, 25, 30, 35, 100
])

pearson_kurtosis = kurtosis(data, fisher=False)
excess_kurtosis = kurtosis(data, fisher=True)

print("Pearson Kurtosis:", pearson_kurtosis)
print("Excess Kurtosis:", excess_kurtosis)
```

### 8.Empirical Rule – 68–95–99.7

The Empirical Rule, also called the 68–95–99.7 rule, describes how data is spread around the mean in an approximately normal (bell-shaped and symmetric) distribution.

- About 68% of observations lie within 1 standard deviation of the mean.
- About 95% lie within 2 standard deviations.
- About 99.7% lie within 3 standard deviations.

It states:
    
- 68% of values lie within μ±1σ
- 95% of values lie within μ±2σ
- 99.7% of values lie within μ±3σ

where:
    
μ = mean
σ = standard deviation



```python
import numpy as np

mean = 70
std = 10

one_sd = (mean - std, mean + std)
two_sd = (mean - 2 * std, mean + 2 * std)
three_sd = (mean - 3 * std, mean + 3 * std)

print("Within 1 SD:", one_sd)
print("Within 2 SD:", two_sd)
print("Within 3 SD:", three_sd)
```

```python
import numpy as np

mean = 70
std = 10

# Generate normally distributed scores
np.random.seed(42)
scores = np.random.normal(mean, std, 10000)

within_1_std = np.mean((scores >= mean - std) & (scores <= mean + std))
within_2_std = np.mean((scores >= mean - 2 * std) & (scores <= mean + 2 * std))
within_3_std = np.mean((scores >= mean - 3 * std) & (scores <= mean + 3 * std))

print("Within 1 SD:", within_1_std * 100, "%")
print("Within 2 SD:", within_2_std * 100, "%")
print("Within 3 SD:", within_3_std * 100, "%")
```
